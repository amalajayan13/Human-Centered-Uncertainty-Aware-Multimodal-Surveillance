## 3. Statistical utilities  **[NEW]**

BCa bootstrap intervals, paired comparison with Holm correction, and expected calibration error.

```python
# ============================================================
# 3. Statistical utilities  [NEW]
# ============================================================
def bca_ci(x, alpha=0.05, n_boot=2000, seed=42):
    """Bias-corrected and accelerated bootstrap CI for the mean."""
    x = np.asarray(x, dtype=float); n = len(x)
    if n < 2 or np.allclose(x, x[0]): return float(x.mean()), float(x.mean()), float(x.mean())
    rng = np.random.default_rng(seed)
    boot = np.array([rng.choice(x, n, replace=True).mean() for _ in range(n_boot)])
    theta = x.mean()
    z0 = stats.norm.ppf(np.clip((boot < theta).mean(), 1e-6, 1-1e-6))
    jack = np.array([np.delete(x, i).mean() for i in range(n)])
    jm = jack.mean(); num = ((jm - jack)**3).sum(); den = 6.0*(((jm - jack)**2).sum()**1.5)
    a = 0.0 if den == 0 else num/den
    def endpoint(q):
        zq = stats.norm.ppf(q)
        adj = z0 + (z0 + zq)/(1 - a*(z0 + zq))
        return float(np.percentile(boot, 100*stats.norm.cdf(adj)))
    return float(theta), endpoint(alpha/2), endpoint(1-alpha/2)

def paired_compare(a, b, label_a="A", label_b="B"):
    """Paired difference a-b with BCa CI and Wilcoxon signed-rank p-value."""
    a = np.asarray(a, float); b = np.asarray(b, float); d = a - b
    m, lo, hi = bca_ci(d)
    try:    p = stats.wilcoxon(a, b).pvalue
    except Exception: p = np.nan
    return dict(comparison=f"{label_a} - {label_b}", mean_diff=m, ci_lo=lo, ci_hi=hi,
                p_value=p, significant=bool(lo > 0 or hi < 0))

def holm(pvals):
    """Holm-Bonferroni adjusted p-values."""
    p = np.asarray(pvals, float); n = len(p); order = np.argsort(p)
    adj = np.empty(n); run = 0.0
    for rank, idx in enumerate(order):
        run = max(run, (n - rank) * p[idx]); adj[idx] = min(run, 1.0)
    return adj

def expected_calibration_error(y_true, y_score, n_bins=15):
    bins = np.linspace(0, 1, n_bins + 1); idx = np.digitize(y_score, bins) - 1
    ece = 0.0
    for b in range(n_bins):
        m = idx == b
        if m.sum() == 0: continue
        ece += (m.sum()/len(y_score)) * abs(y_true[m].mean() - y_score[m].mean())
    return float(ece)

METRIC_KEYS = ["accuracy","bal_acc","precision","recall","f1","roc_auc","pr_auc","fpr"]

def compute_metrics(y_true, y_pred, y_score, minority_class=None):
    y_true = np.asarray(y_true).astype(int); y_pred = np.asarray(y_pred).astype(int)
    tn, fp, fn, tp = confusion_matrix(y_true, y_pred, labels=[0,1]).ravel()
    out = dict(accuracy=accuracy_score(y_true,y_pred), bal_acc=balanced_accuracy_score(y_true,y_pred),
        precision=precision_score(y_true,y_pred,zero_division=0),
        recall=recall_score(y_true,y_pred,zero_division=0),
        f1=f1_score(y_true,y_pred,zero_division=0),
        roc_auc=roc_auc_score(y_true,y_score) if len(np.unique(y_true))>1 else np.nan,
        pr_auc=average_precision_score(y_true,y_score) if len(np.unique(y_true))>1 else np.nan,
        fpr=fp/(fp+tn) if (fp+tn)>0 else np.nan,
        tn=int(tn), fp=int(fp), fn=int(fn), tp=int(tp))
    if minority_class is not None:   # metrics on the stated class
        out["precision_minority"] = precision_score(y_true,y_pred,pos_label=minority_class,zero_division=0)
        out["recall_minority"]    = recall_score(y_true,y_pred,pos_label=minority_class,zero_division=0)
        out["f1_minority"]        = f1_score(y_true,y_pred,pos_label=minority_class,zero_division=0)
    return out

def summarise_reps(df, group_cols, metric_keys=METRIC_KEYS, alpha=0.05, n_boot=2000):
    """mean + 95% BCa CI for each metric, over repetitions."""
    rows = []
    for keys, g in df.groupby(group_cols):
        keys = keys if isinstance(keys, tuple) else (keys,)
        row = dict(zip(group_cols, keys))
        for k in metric_keys:
            if k not in g: continue
            m, lo, hi = bca_ci(g[k].dropna().values, alpha, n_boot, CONFIG["random_state"])
            row[k], row[f"{k}_lo"], row[f"{k}_hi"] = m, lo, hi
        row["n_reps"] = len(g); rows.append(row)
    return pd.DataFrame(rows)

def fmt_ci(m, lo, hi, d=3):
    return f"{m:.{d}f} [{lo:.{d}f}, {hi:.{d}f}]"

def to_latex(df, name, caption, label, col_format=None, escape=True):
    """Write a LaTeX tabular fragment ready to \input into the manuscript."""
    body = df.to_latex(index=False, escape=escape, column_format=col_format,
                       float_format=lambda v: f"{v:.3f}")
    tex = ("% auto-generated -- do not edit by hand\n"
           "\\begin{table*}[!t]\n\\centering\n\\small\n"
           f"\\caption{{{caption}}}\n\\label{{{label}}}\n{body}\\end{{table*}}\n")
    (LATEX_PATH / name).write_text(tex, encoding="utf-8")
    log(f"saved latex/{name}")
```

## 4. Surveillance-event generator

Unchanged from v2, so the stream is bit-identical to the submitted one (43,200 events;
hidden-incident 34,819/8,381; label 20,577/22,623). The generator is reproduced here in full
because Section 3.3.1 of the manuscript documents it.

```python
# ============================================================
# 4. Surveillance-event generator  (unchanged -- reproduces the submitted stream)
# ============================================================
def generate_realistic_video_events(config, rng):
    n_cameras = int(config["video_n_cameras"]); n_zones = int(config["video_n_zones"])
    days = int(config["video_days"]); freq = config["video_event_freq"]
    start = pd.Timestamp("2024-01-01 00:00:00")
    ts = pd.date_range(start=start, end=start + pd.Timedelta(days=days) - pd.to_timedelta(freq), freq=freq)
    n = len(ts)
    camera_ids = rng.integers(0, n_cameras, size=n)
    zone_ids   = rng.integers(0, n_zones,   size=n)

    # latent incident state: first-order Markov chain with burst persistence
    hidden = np.zeros(n, dtype=int)
    base, burst = float(config["video_hidden_incident_rate"]), float(config["video_correlated_burst_prob"])
    for i in range(n):
        p = base if i == 0 else min(max(base + (burst if hidden[i-1] == 1 else 0.0), 0.0), 0.95)
        hidden[i] = rng.binomial(1, p)

    camera_risk = rng.normal(0.0, 0.35, size=n_cameras)
    zone_risk   = rng.normal(0.0, 0.40, size=n_zones)

    motion   = np.clip(0.35 + 0.35*hidden + camera_risk[camera_ids]*0.15 + rng.normal(0,0.18,n), 0, 1)
    density  = np.clip(0.30 + 0.30*hidden + zone_risk[zone_ids]*0.20  + rng.normal(0,0.20,n), 0, 1)
    auth     = np.clip(0.65 - 0.15*hidden + rng.normal(0,0.20,n), 0, 1)
    cons     = np.clip(0.60 - 0.10*hidden + rng.normal(0,0.22,n), 0, 1)
    unc      = np.clip(0.35 + 0.20*hidden + rng.normal(0,0.20,n), 0, 1)

    logit = (1.2*hidden + 1.1*motion + 0.9*density - 0.6*auth - 0.4*cons + 0.6*unc
             + 0.20*camera_risk[camera_ids] + 0.25*zone_risk[zone_ids] + rng.normal(0,0.65,n))
    prob  = 1/(1+np.exp(-logit))
    label = (prob > 0.58).astype(int)
    flip  = rng.random(n) < float(config["video_label_flip_prob"])
    label[flip] = 1 - label[flip]      # 4% label noise floor

    df = pd.DataFrame(dict(timestamp=ts, camera_id=camera_ids, zone_id=zone_ids,
        motion_intensity=motion, scene_density=density, authenticity_score=auth,
        temporal_consistency=cons, uncertainty_score=unc, hidden_incident=hidden,
        suspicious_probability=prob, suspicious_label=label))
    meta = dict(rows=int(len(df)), n_cameras=n_cameras, n_zones=n_zones,
        start=str(df.timestamp.min()), end=str(df.timestamp.max()),
        label_distribution=df.suspicious_label.value_counts().sort_index().to_dict(),
        hidden_distribution=df.hidden_incident.value_counts().sort_index().to_dict())
    return df, meta

section("Generate surveillance-event stream")
rng = np.random.default_rng(CONFIG["random_state"])
video_df, video_meta = generate_realistic_video_events(CONFIG, rng)
video_df.to_csv(RUN_DIR / "video_events_simulated_realistic.csv", index=False)
log(json.dumps(video_meta, indent=2)); save_json(video_meta, "video_generation_meta.json")

assert video_meta["rows"] == 43200
assert video_meta["hidden_distribution"] == {0: 34819, 1: 8381}, "generator drift -- stream no longer matches the submission"
assert video_meta["label_distribution"]  == {0: 20577, 1: 22623}, "generator drift -- stream no longer matches the submission"
log("\\nOK: regenerated stream is identical to the submitted one.")
```

## 7. Splitting and repeated evaluation  **[CHANGED C2 / NEW C3]**

Three changes:

* **Validation partition.** 60/10/30. Every threshold, window size and coefficient is chosen on
  validation; the test partition is scored once per repetition.
* **Repeated evaluation.** The surveillance stream is temporally ordered and burst-correlated, so
  an i.i.d. resample would leak across the split. We use **rolling-origin** repetitions: `n_repetitions`
  cut points sliding across the timeline, each giving a disjoint train/val/test triple. The cyber
  branch, which has no genuine temporal structure, uses repeated stratified resampling.
* **Alternative holdouts.** Chronological (default), stratified, and camera-level group holdout.

```python
# ============================================================
# 7. Split generators  [CHANGED/NEW]
# ============================================================
def rolling_origin_splits(n, n_reps, train_frac, val_frac, test_frac):
    """Disjoint train/val/test index triples at n_reps sliding cut points."""
    out = []
    span = 1.0 - (train_frac + val_frac + test_frac)     # slack to slide over
    starts = np.linspace(train_frac, train_frac + max(span, 0.0) + 0.19, n_reps)
    for f in starts:
        cut = int(min(f, 0.79) * n)
        v_end = int(cut + val_frac * n)
        if v_end >= n - 10: continue
        out.append((np.arange(0, cut), np.arange(cut, v_end), np.arange(v_end, n)))
    return out

def stratified_splits(y, n_reps, train_frac, val_frac, seed=42):
    out = []
    idx = np.arange(len(y))
    for r in range(n_reps):
        tr_va, te = train_test_split(idx, test_size=1-(train_frac+val_frac),
                                     stratify=y, random_state=seed+r)
        tr, va = train_test_split(tr_va, test_size=val_frac/(train_frac+val_frac),
                                  stratify=np.asarray(y)[tr_va], random_state=seed+r)
        out.append((tr, va, te))
    return out

def group_holdout_splits(groups, y, n_reps, seed=42):
    """Hold out whole camera identities: no camera appears in both train and test."""
    out = []; uniq = np.unique(groups); rng = np.random.default_rng(seed)
    n_test_g = max(1, int(round(0.3*len(uniq))))
    n_val_g  = max(1, int(round(0.1*len(uniq))))
    for r in range(n_reps):
        perm = rng.permutation(uniq)
        te_g, va_g = perm[:n_test_g], perm[n_test_g:n_test_g+n_val_g]
        te = np.where(np.isin(groups, te_g))[0]
        va = np.where(np.isin(groups, va_g))[0]
        tr = np.where(~np.isin(groups, np.concatenate([te_g, va_g])))[0]
        if len(np.unique(y[te])) < 2 or len(np.unique(y[va])) < 2: continue
        out.append((tr, va, te))
    return out

def make_model(name):
    s = CONFIG["random_state"]
    if name == "LogisticRegression":
        return LogisticRegression(max_iter=1000, random_state=s)
    if name == "CalibratedLogisticRegression":
        return CalibratedClassifierCV(LogisticRegression(max_iter=1000, random_state=s),
                                      method="sigmoid", cv=3)
    if name == "RandomForest":
        return RandomForestClassifier(n_estimators=CONFIG["n_estimators"], max_depth=CONFIG["max_depth"],
            min_samples_leaf=CONFIG["min_samples_leaf"], class_weight=CONFIG["class_weight"],
            random_state=s, n_jobs=-1)
    if name == "ExtraTrees":
        return ExtraTreesClassifier(n_estimators=CONFIG["n_estimators"], max_depth=CONFIG["max_depth"],
            min_samples_leaf=CONFIG["min_samples_leaf"], class_weight=CONFIG["class_weight"],
            random_state=s, n_jobs=-1)
    raise ValueError(name)

def video_pipeline(cols, model_name):
    return Pipeline([("pre", ColumnTransformer([("num", Pipeline([
        ("imp", SimpleImputer(strategy="median")), ("sc", StandardScaler())]), cols)],
        remainder="drop")), ("model", make_model(model_name))])

def cyber_pipeline(model_name):
    return Pipeline([("pre", ColumnTransformer([
        ("num", Pipeline([("imp", SimpleImputer(strategy="median")), ("sc", StandardScaler())]), cyber_numeric),
        ("cat", Pipeline([("imp", SimpleImputer(strategy="most_frequent")),
                          ("oh", OneHotEncoder(handle_unknown="ignore"))]), cyber_categorical)],
        remainder="drop")), ("model", make_model(model_name))])

def select_threshold(y_val, s_val, objective="bal_acc"):
    """Threshold chosen on VALIDATION only.  [CHANGED C2]"""
    best, best_t = -np.inf, 0.5
    for t in CONFIG["threshold_grid"]:
        pred = (s_val >= t).astype(int)
        v = balanced_accuracy_score(y_val, pred) if objective == "bal_acc" else f1_score(y_val, pred, zero_division=0)
        if v > best: best, best_t = v, t
    return float(best_t)
```

## 11. UVIF operator  **[CORRECTED C3]**

Stated exactly as implemented, matching manuscript Eqs. (8)–(14):

* confidence mass `mu = 1 - u`, **upper cap only** at 0.80
* weights `w = mu / sum(mu)` (this *is* inverse-uncertainty weighting; lambda = 0)
* **subtractive** penalty: `s_adj = clip(s_raw - 0.30 * u_fusion)`

Also reports the fraction of windows where **both** branches hit the cap — there UVIF is
*identical* to unweighted mean fusion.

```python
# ============================================================
# 11. UVIF operator and fusion baselines  [CORRECTED / NEW]
# ============================================================
def uncertainty_proxy(p):
    return 1.0 - 2.0*np.abs(np.asarray(p, float) - 0.5)

def uvif_weights(u_c, u_v, cap=None):
    cap = CONFIG["fusion_confidence_cap"] if cap is None else cap
    mu_c = np.minimum(1.0 - np.asarray(u_c, float), cap)      # upper cap only
    mu_v = np.minimum(1.0 - np.asarray(u_v, float), cap)
    s = mu_c + mu_v; s = np.where(s == 0, 1e-9, s)
    return mu_c/s, mu_v/s

def uvif_score(p_c, p_v, u_c, u_v, kappa=None, cap=None):
    kappa = CONFIG["fusion_uncertainty_penalty"] if kappa is None else kappa
    w_c, w_v = uvif_weights(u_c, u_v, cap)
    raw = w_c*np.asarray(p_c,float) + w_v*np.asarray(p_v,float)
    u_f = w_c*np.asarray(u_c,float) + w_v*np.asarray(u_v,float)
    adj = np.clip(raw - kappa*u_f, 0, 1)              # SUBTRACTIVE  [CORRECTED]
    return adj, raw, u_f, w_c, w_v

# ---------- baselines (identical inputs, only the rule differs) ----------
def fuse_mean(p_c, p_v, **k):  return 0.5*(np.asarray(p_c)+np.asarray(p_v))
def fuse_max(p_c, p_v, **k):   return np.maximum(p_c, p_v)
def fuse_fixed(p_c, p_v, w=0.5, **k): return w*np.asarray(p_c) + (1-w)*np.asarray(p_v)
def fuse_confidence(p_c, p_v, **k):
    """confidence weighting WITHOUT the variational formulation and WITHOUT capping"""
    c_c = np.abs(np.asarray(p_c)-0.5); c_v = np.abs(np.asarray(p_v)-0.5)
    s = c_c + c_v; s = np.where(s == 0, 1e-9, s)
    return (c_c*np.asarray(p_c) + c_v*np.asarray(p_v))/s
def fuse_dempster_shafer(p_c, p_v, **k):
    """two simple support functions; conflict mass renormalised"""
    p_c = np.clip(p_c, 1e-6, 1-1e-6); p_v = np.clip(p_v, 1e-6, 1-1e-6)
    m1 = p_c*p_v                       # both support "attack"
    m0 = (1-p_c)*(1-p_v)               # both support "benign"
    conflict = 1 - m1 - m0
    return m1/np.clip(1-conflict, 1e-9, None)
def fuse_evidential(p_c, p_v, **k):
    """established uncertainty-aware reference: evidence-strength weighting (Shao et al. style)"""
    e_c = np.abs(np.asarray(p_c)-0.5)*2; e_v = np.abs(np.asarray(p_v)-0.5)*2
    s = e_c + e_v + 2.0                                     # +2 = vacuous prior mass
    return (e_c*np.asarray(p_c) + e_v*np.asarray(p_v) + 1.0)/s

FUSION_RULES = {
    "B1 Mean":                    fuse_mean,
    "B2 Maximum":                 fuse_max,
    "B3 Fixed weight":            fuse_fixed,
    "B4 Confidence (no variational)": fuse_confidence,
    "B5 Dempster-Shafer":         fuse_dempster_shafer,
    "B6 Evidential [Shao2023]":   fuse_evidential,
}
log("UVIF and 6 baselines defined.")
```

## 12. Fusion stage

Window construction is stated accurately: events are binned into 120 s windows on **two
independently synthesised clocks** and the window tables are outer-joined. There is **no
record-level pairing** — the 12,961 rows are windows.

```python
# ============================================================
# 12. Fusion  -> tab:fusion_metrics_or / tab:fusion_baselines
# ============================================================
def assign_window_id(ts, origin, seconds):
    return ((ts - origin).dt.total_seconds() // seconds).astype(int)

def build_event_tables(cyber_scores, cyber_labels, video_scores, video_labels,
                       video_timestamps, ordering="native", seed=42):
    """Cyber timestamps are SYNTHETIC (surrogate ordering); this is stated in the manuscript."""
    n_c = len(cyber_scores)
    if ordering == "random":
        perm = np.random.default_rng(seed).permutation(n_c)
    elif ordering == "rate_stratified":
        # interleave positives and negatives so that the per-window malicious rate matches the global rate
        pos = np.where(np.asarray(cyber_labels) == 1)[0]
        neg = np.where(np.asarray(cyber_labels) == 0)[0]
        rng = np.random.default_rng(seed); rng.shuffle(pos); rng.shuffle(neg)
        ratio = max(1, int(round(len(pos)/max(1, len(neg)))))
        perm, pi, ni = [], 0, 0
        while pi < len(pos) or ni < len(neg):
            for _ in range(ratio):
                if pi < len(pos): perm.append(pos[pi]); pi += 1
            if ni < len(neg): perm.append(neg[ni]); ni += 1
        perm = np.asarray(perm)
    else:
        perm = np.arange(n_c)
    cs, cl = np.asarray(cyber_scores)[perm], np.asarray(cyber_labels)[perm]
    # 20 s spacing is the value used in the submitted run; it is what makes the synthetic cyber
    # range overlap the surveillance test range and yields the reported 12,961 windows.
    cyber_ts = pd.date_range(start="2024-02-12 00:00:00", periods=n_c, freq="20s")
    cyber_ev = pd.DataFrame(dict(timestamp=cyber_ts, cyber_score=cs, cyber_label=cl,
                                 cyber_uncertainty=uncertainty_proxy(cs)))
    video_ev = pd.DataFrame(dict(timestamp=pd.to_datetime(video_timestamps),
                                 video_score=video_scores, video_label=video_labels,
                                 video_uncertainty=uncertainty_proxy(video_scores)))
    return cyber_ev, video_ev

def build_fusion_table(cyber_ev, video_ev, window_seconds, tau_c, tau_v, kappa=None, cap=None):
    origin = min(cyber_ev.timestamp.min(), video_ev.timestamp.min())
    cyber_ev = cyber_ev.copy(); video_ev = video_ev.copy()
    cyber_ev["window_id"] = assign_window_id(cyber_ev.timestamp, origin, window_seconds)
    video_ev["window_id"] = assign_window_id(video_ev.timestamp, origin, window_seconds)
    cyber_ev["cyber_pred"] = (cyber_ev.cyber_score >= tau_c).astype(int)
    video_ev["video_pred"] = (video_ev.video_score >= tau_v).astype(int)

    cw = cyber_ev.groupby("window_id").agg(cyber_mean_prob=("cyber_score","mean"),
        cyber_mean_uncertainty=("cyber_uncertainty","mean"), cyber_count=("cyber_score","size"),
        cyber_label_any=("cyber_label","max")).reset_index()
    vw = video_ev.groupby("window_id").agg(video_mean_prob=("video_score","mean"),
        video_mean_uncertainty=("video_uncertainty","mean"), video_count=("video_score","size"),
        video_label_any=("video_label","max")).reset_index()

    t = cw.merge(vw, on="window_id", how="outer").fillna(0)      # OUTER JOIN over windows
    t["modality_count"] = (t.cyber_count>0).astype(int) + (t.video_count>0).astype(int)

    adj, raw, u_f, w_c, w_v = uvif_score(t.cyber_mean_prob, t.video_mean_prob,
                                         t.cyber_mean_uncertainty, t.video_mean_uncertainty, kappa, cap)
    t["w_cyber"], t["w_video"] = w_c, w_v
    t["raw_fusion_score"], t["fusion_uncertainty"], t["adjusted_fusion_score"] = raw, u_f, adj

    # three fusion targets  [NEW]
    t["target_or"]  = ((t.cyber_label_any>0) | (t.video_label_any>0)).astype(int)
    t["target_and"] = ((t.cyber_label_any>0) & (t.video_label_any>0)).astype(int)
    t["target_consensus"] = (0.5*t.cyber_label_any + 0.5*t.video_label_any >= 0.5).astype(int)
    # disagreement subset: the only rows where a combination rule can express a preference  [NEW]
    t["disagree"] = ((t.cyber_label_any>0) != (t.video_label_any>0)) & (t.modality_count==2)
    # cap-saturation diagnostic  [NEW]
    cap_v = CONFIG["fusion_confidence_cap"] if cap is None else cap
    t["both_capped"] = ((1-t.cyber_mean_uncertainty >= cap_v) & (1-t.video_mean_uncertainty >= cap_v))
    return t, cyber_ev, video_ev

def window_state(row, theta):
    if row.modality_count == 1: return "single_modality_only"
    if row.adjusted_fusion_score >= theta: return "high_risk"
    if row.fusion_uncertainty >= CONFIG["window_uncertainty_threshold"]: return "uncertain"
    return "low_risk" if row.adjusted_fusion_score > 0 else "benign"
log("Fusion machinery defined.")
```

### 12b. Run the fusion stage and the baseline comparison  **[NEW C4]**

All rules see identical branch posteriors, identical windows and identical evaluation data, so the
combination rule is the only difference. Reported three ways: whole test set, **cyber-only
reference row**, and the **disagreement subset**.

```python
# ============================================================
# 12b. Fusion results and baselines  -> tab:fusion_baselines
# ============================================================
if not CYBER_AVAILABLE:
    log("SKIPPED: fusion requires the cyber branch.")
else:
    section("Fusion stage")
    # branch models for fusion, refit on a single train/val/test split of each branch
    ctr, cva, cte = cyber_splits[0]
    vsp = splits_chrono[0]; vtr, vva, vte = vsp

    fusion_cyber = cyber_pipeline(CONFIG["force_fusion_cyber_model"]).fit(Xc.iloc[ctr], yc.iloc[ctr])
    # the operational surveillance model is the best branch model, not a fixed choice
    _best = (video_summary[video_summary.feature_set == "audited_full"]
             .sort_values("bal_acc", ascending=False).model.iloc[0])
    log(f"operational surveillance model selected from branch results: {_best}")
    fusion_video = video_pipeline(video_full_feature_cols, _best).fit(
        video_df[video_full_feature_cols].iloc[vtr], yv.iloc[vtr])

    # branch alert thresholds: selected on VALIDATION  [CHANGED C2]
    tau_c = select_threshold(yc.iloc[cva].values, fusion_cyber.predict_proba(Xc.iloc[cva])[:,1])
    tau_v = select_threshold(yv.iloc[vva].values,
                             fusion_video.predict_proba(video_df[video_full_feature_cols].iloc[vva])[:,1])
    log(f"validation-selected alert thresholds: tau_c={tau_c}, tau_v={tau_v}")

    c_s = fusion_cyber.predict_proba(Xc.iloc[cte])[:,1];  c_y = yc.iloc[cte].values
    v_s = fusion_video.predict_proba(video_df[video_full_feature_cols].iloc[vte])[:,1]
    v_y = yv.iloc[vte].values; v_ts = video_df.timestamp.iloc[vte].values

    cyber_ev, video_ev = build_event_tables(c_s, c_y, v_s, v_y, v_ts, ordering="native")
    ftab, cyber_ev, video_ev = build_fusion_table(cyber_ev, video_ev,
                                CONFIG["fusion_window_seconds"], tau_c, tau_v)

    # Threshold on an INTERLEAVED validation half of the windows.  [CHANGED C2]
    # A first-half/second-half split would be wrong here: the two synthetic clocks overlap only on a
    # sub-range, so a contiguous second half contains no dual-modality windows at all.
    ftab = ftab.sort_values("window_id").reset_index(drop=True)
    val_mask = (np.arange(len(ftab)) % 2 == 0); test_mask = ~val_mask
    theta = select_threshold(ftab.target_or.values[val_mask], ftab.adjusted_fusion_score.values[val_mask])
    ftab["fused_pred"] = (ftab.adjusted_fusion_score >= theta).astype(int)
    ftab["window_state"] = ftab.apply(lambda r: window_state(r, theta), axis=1)
    save_csv(ftab, "fusion_window_table.csv")
    log(f"fusion rows (WINDOWS, not paired records): {len(ftab)}")
    # reproduction guards against the submitted run
    for name, got, want in [("fusion windows", len(ftab), 12961),
                            ("dual-modality windows", int((ftab.modality_count==2).sum()), 9524),
                            ("single-modality windows", int((ftab.modality_count==1).sum()), 3437)]:
        log(f"  check {name}: {got} (submitted run: {want})" + ("" if got==want else "   <-- DIFFERS"))
    log(f"validation-selected fusion threshold theta = {theta}")
    log(f"windows where BOTH branches hit the confidence cap "
        f"(UVIF == mean fusion there): {ftab.both_capped.mean():.3%}")
    log(f"window states: {ftab.window_state.value_counts().to_dict()}")
    save_json({"fusion_rows": int(len(ftab)), "theta": theta, "tau_c": tau_c, "tau_v": tau_v,
               "both_capped_fraction": float(ftab.both_capped.mean()),
               "window_states": ftab.window_state.value_counts().to_dict(),
               "dual_modality_windows": int((ftab.modality_count==2).sum()),
               "single_modality_windows": int((ftab.modality_count==1).sum())}, "fusion_summary.json")

    # ---------- baseline comparison ----------
    eval_test = ftab[test_mask].copy()           # untouched interleaved half
    def eval_rule(scores, target, mask=None):
        sc = np.asarray(scores); tg = np.asarray(target)
        if mask is not None: sc, tg = sc[mask], tg[mask]
        if len(np.unique(tg)) < 2: return None
        t = select_threshold(tg, sc)
        return compute_metrics(tg, (sc>=t).astype(int), sc)

    def block_bootstrap(n, B=800, block=50, seed=42):
        rng = np.random.default_rng(seed); nb = int(np.ceil(n/block))
        return [np.concatenate([np.arange(s, min(s+block, n))
                                for s in rng.integers(0, max(1, n-block), nb)])[:n] for _ in range(B)]

    rows = []
    p_c_w, p_v_w = eval_test.cyber_mean_prob.values, eval_test.video_mean_prob.values
    u_c_w, u_v_w = eval_test.cyber_mean_uncertainty.values, eval_test.video_mean_uncertainty.values
    uvif_w = uvif_score(p_c_w, p_v_w, u_c_w, u_v_w)[0]

    # thresholds are taken from the validation half, never from the evaluation half
    v = ftab[val_mask]
    uvif_v = uvif_score(v.cyber_mean_prob.values, v.video_mean_prob.values,
                        v.cyber_mean_uncertainty.values, v.video_mean_uncertainty.values)[0]
    RULES_T = {"Cyber branch only (reference)": (p_c_w, v.cyber_mean_prob.values)}
    for rname, fn in FUSION_RULES.items():
        RULES_T[rname] = (fn(p_c_w, p_v_w), fn(v.cyber_mean_prob.values, v.video_mean_prob.values))
    RULES_T["UVIF (proposed)"] = (uvif_w, uvif_v)

    def evaluate_family(mask, label, target="target_or", B=800):
        y = eval_test[target].values
        y = y if mask is None else y[mask]
        if len(np.unique(y)) < 2:
            log(f"!! {label}: target is constant ({sorted(set(y))}) -- no rule can be "
                f"rewarded or penalised on this subset. Reported as degenerate.")
            return []
        boots = block_bootstrap(len(y), B)
        res = {}
        for rname, (s_te, s_va) in RULES_T.items():
            s_ = s_te if mask is None else s_te[mask]
            thr = select_threshold(v[target].values, s_va)
            ba, au = [], []
            for bi in boots:
                yy, ss = y[bi], s_[bi]
                if len(np.unique(yy)) < 2: continue
                ba.append(balanced_accuracy_score(yy, (ss >= thr).astype(int)))
                au.append(roc_auc_score(yy, ss))
            res[rname] = dict(m=compute_metrics(y, (s_ >= thr).astype(int), s_),
                              thr=thr, ba=np.array(ba), au=np.array(au))
        out = []
        ub, ua = res["UVIF (proposed)"]["ba"], res["UVIF (proposed)"]["au"]
        for rname, r in res.items():
            db, da = ub - r["ba"], ua - r["au"]
            row = dict(rule=rname, subset=label, target=target, threshold=r["thr"], **r["m"])
            row["bal_acc_lo"], row["bal_acc_hi"] = np.percentile(r["ba"], [2.5, 97.5])
            row["roc_auc_lo"], row["roc_auc_hi"] = np.percentile(r["au"], [2.5, 97.5])
            if rname != "UVIF (proposed)":
                row["d_bal"] = f"{db.mean():+.4f} [{np.percentile(db,2.5):+.4f}, {np.percentile(db,97.5):+.4f}]"
                row["d_auc"] = f"{da.mean():+.4f} [{np.percentile(da,2.5):+.4f}, {np.percentile(da,97.5):+.4f}]"
            else:
                row["d_bal"] = row["d_auc"] = "---"
            out.append(row)
        return out

    rows += evaluate_family(None, "All windows", "target_or")
    rows += evaluate_family(None, "All windows", "target_and")

    # The disagreement subset is the only subset on which a combination rule can express a
    # preference. Check whether it is degenerate -- in the submitted configuration it is.
    dis = eval_test.disagree.values.astype(bool)
    log(f"\ndisagreement windows in the evaluation half: {int(dis.sum())}")
    if dis.sum() > 20:
        cy = eval_test.cyber_label_any.values[dis]; vi = eval_test.video_label_any.values[dis]
        log(f"  cyber label on that subset: {dict(zip(*np.unique(cy, return_counts=True)))}")
        log(f"  surveillance label on that subset: {dict(zip(*np.unique(vi, return_counts=True)))}")
        eval_test["target_video"] = (eval_test.video_label_any > 0).astype(int)
        rows += evaluate_family(dis, "Disagreement subset", "target_or")
        rows += evaluate_family(dis, "Disagreement subset", "target_video")

    baselines = pd.DataFrame(rows)
    save_csv(baselines, "fusion_baselines_raw.csv")
    show = baselines[(baselines.target=="target_or") & (baselines.subset=="All windows")][
        ["rule","threshold","bal_acc","bal_acc_lo","bal_acc_hi","roc_auc","roc_auc_lo","roc_auc_hi","fpr","d_bal"]]
    to_latex(show, "tab_fusion_baselines.tex",
        "Fusion-baseline comparison under identical branch posteriors, identical windows and "
        "identical evaluation data. The cyber-only row is the reference against which any claim of "
        "multimodal gain must be read; the disagreement subset is the only subset on which a "
        "combination rule can express a preference.", "tab:fusion_baselines")
    display(show)
```

## 14. Graph construction  **[CORRECTED C5]**

Documented exactly as implemented, plus the parameter sweep. Note in particular that the
similarity rule is **within-modality only** and that cosine similarity on non-negative 2-D
`(score, uncertainty)` vectors is satisfied by nearly every pair at 0.92 — which is what produces
the giant component. A Euclidean alternative is provided.

```python
# ============================================================
# 14. Heterogeneous temporal graph  [CORRECTED C5; sparse]
# ============================================================
# NOTE: the deployed configuration yields ~19.5M edges over ~63k nodes. A networkx
# Graph at that size needs several GB, so components are computed with scipy sparse.
def build_edges(ca, va, ftab, eps=None, cap=None, temporal=None, metric=None):
    eps      = CONFIG["graph_similarity_threshold"]   if eps is None else eps
    cap      = CONFIG["graph_max_similarity_sample"]  if cap is None else cap
    temporal = CONFIG["graph_temporal_edge_seconds"]  if temporal is None else temporal
    metric   = CONFIG["graph_similarity_metric"]      if metric is None else metric
    nC, nV = len(ca), len(va)
    hub_w = ftab[ftab.modality_count >= 1].window_id.values      # hub per window with >=1 alert
    hub_idx = {w: nC + nV + i for i, w in enumerate(hub_w)}
    N = nC + nV + len(hub_w)
    rows, cols, fam = [], [], {"window": 0, "similarity": 0, "temporal": 0}

    for off, df, wcol in ((0, ca, "window_id"), (nC, va, "window_id")):
        for i, w in enumerate(df[wcol].values):
            h = hub_idx.get(int(w))
            if h is not None:
                rows.append(off + i); cols.append(h); fam["window"] += 1

    # similarity edges -- WITHIN MODALITY ONLY, on the (score, uncertainty) vector
    for off, df, sc, uc in ((0, ca, "cyber_score", "cyber_uncertainty"),
                            (nC, va, "video_score", "video_uncertainty")):
        a = df.iloc[:cap]
        if len(a) < 2: continue
        V = a[[sc, uc]].values.astype(float)
        iu = np.triu_indices(len(a), 1)
        if metric == "cosine":
            n = np.linalg.norm(V, axis=1, keepdims=True); n[n == 0] = 1e-9
            S = (V/n) @ (V/n).T; hit = S[iu] >= eps
        else:                                     # euclidean alternative
            D = np.linalg.norm(V[:, None, :] - V[None, :, :], axis=-1); hit = D[iu] <= (1 - eps)
        rows.extend((iu[0][hit] + off).tolist()); cols.extend((iu[1][hit] + off).tolist())
        fam["similarity"] += int(hit.sum())

    # short-range temporal chain edges between consecutive same-modality alerts
    if temporal > 0:
        for off, df in ((0, ca), (nC, va)):
            a = df.sort_values("timestamp").reset_index(drop=True)
            dt = a.timestamp.diff().dt.total_seconds().values[1:]
            k = np.where(dt <= temporal)[0]
            rows.extend((k + off).tolist()); cols.extend((k + 1 + off).tolist())
            fam["temporal"] += len(k)
    return np.asarray(rows), np.asarray(cols), N, nC, nV, fam

def graph_stats(r, c, N):
    if len(r) == 0:
        return dict(nodes=N, edges=0, components=N, largest=1, largest_frac=1/N,
                    singletons=N, non_singleton=0), None
    A = coo_matrix((np.ones(len(r)), (r, c)), shape=(N, N))
    ncomp, lab = connected_components(A, directed=False)
    cnt = np.bincount(lab)
    return dict(nodes=N, edges=int(len(r)), components=int(ncomp), largest=int(cnt.max()),
                largest_frac=float(cnt.max()/N), singletons=int((cnt == 1).sum()),
                non_singleton=int((cnt > 1).sum())), (lab, cnt)

if CYBER_AVAILABLE:
    section("Graph construction")
    ca = cyber_ev[cyber_ev.cyber_pred == 1].reset_index(drop=True)
    va_ = video_ev[video_ev.video_pred == 1].reset_index(drop=True)
    log(f"retained alerts: cyber={len(ca)}, surveillance={len(va_)}")
    r_, c_, N, nC, nV, fam = build_edges(ca, va_, ftab)
    gs, (lab, cnt) = graph_stats(r_, c_, N)
    gs["edge_families"] = fam
    gs["direct_cross_modal_edges"] = int(((r_ < nC) & (c_ >= nC) & (c_ < nC+nV)).sum()
                                       + ((c_ < nC) & (r_ >= nC) & (r_ < nC+nV)).sum())
    log(json.dumps(gs, indent=2)); save_json(gs, "graph_summary.json")
    log("NOTE: the similarity rule never crosses modalities, so the giant component is two "
        "same-modality cliques joined at hubs, not a multimodal structure.")

    sizes = np.sort(cnt)[::-1]
    census = pd.DataFrame([
        ("Singletons (1 node)", int((cnt==1).sum()), int(cnt[cnt==1].sum())),
        ("Small (2--42 nodes)", int(((cnt>=2)&(cnt<=42)).sum()), int(cnt[(cnt>=2)&(cnt<=42)].sum())),
        ("Medium (43--199 nodes)", int(((cnt>=43)&(cnt<=199)).sum()), int(cnt[(cnt>=43)&(cnt<=199)].sum())),
        ("Large (>=200, excl. giant)", int((sizes[1:]>=200).sum()), int(sizes[1:][sizes[1:]>=200].sum())),
        ("Giant component", 1, int(sizes[0])),
        ("Total", int(len(cnt)), int(cnt.sum()))],
        columns=["Component class","Count","Nodes covered"])
    census["Share of nodes"] = (census["Nodes covered"]/N).map(lambda v: f"{v:.1%}")
    save_csv(census, "tab_component_summary.csv")
    to_latex(census, "tab_component_summary.tex",
        "Complete connected-component census, including singletons and the giant component.",
        "tab:component_summary")
    display(census)
```

```python
# ============================================================
# 15. Campaign extraction  [CORRECTED C6]  -> tab:campaign_candidates
# ============================================================
if CYBER_AVAILABLE:
    section("Campaign candidates")
    node_score = np.concatenate([ca.cyber_score.values, va_.video_score.values,
                                 ftab[ftab.modality_count>=1].adjusted_fusion_score.values])
    node_unc   = np.concatenate([ca.cyber_uncertainty.values, va_.video_uncertainty.values,
                                 ftab[ftab.modality_count>=1].fusion_uncertainty.values])
    is_cyber = np.arange(N) < nC
    is_video = (np.arange(N) >= nC) & (np.arange(N) < nC+nV)
    is_hub   = np.arange(N) >= nC+nV
    rows = []
    for cid in range(gs["components"]):
        m = lab == cid
        n_ev = int((m & ~is_hub).sum())
        if n_ev < CONFIG["min_campaign_evidence_nodes"]: continue
        rows.append(dict(campaign_id=cid, nodes=int(m.sum()), evidence_nodes=n_ev,
            cyber_nodes=int((m & is_cyber).sum()), video_nodes=int((m & is_video).sum()),
            hub_nodes=int((m & is_hub).sum()),
            mean_score=float(node_score[m & ~is_hub].mean()) if n_ev else np.nan,
            max_score=float(node_score[m & ~is_hub].max()) if n_ev else np.nan,
            mean_fusion=float(node_score[m & is_hub].mean()) if (m & is_hub).any() else np.nan,
            mean_uncertainty=float(node_unc[m].mean())))
    camp = pd.DataFrame(rows)

    # weighted score psi  [NEW: Eq. 29 now actually implemented]
    a,b,g_ = CONFIG["campaign_alpha"], CONFIG["campaign_beta"], CONFIG["campaign_gamma"]
    camp["psi"] = (a*camp.mean_fusion.fillna(camp.mean_score) + b*np.log1p(camp.evidence_nodes)
                   - g_*camp.mean_uncertainty)
    camp_psi = camp.sort_values("psi", ascending=False).reset_index(drop=True)
    camp_lex = camp.sort_values(["nodes","max_score"], ascending=[False,False]).reset_index(drop=True)

    # rank correlation between the deployed lexicographic rule and the weighted rule  [NEW]
    r_psi = camp_psi.campaign_id.tolist(); r_lex = camp_lex.campaign_id.tolist()
    pos = {c:i for i,c in enumerate(r_lex)}
    tau = stats.kendalltau([pos[c] for c in r_psi], list(range(len(r_psi)))) if len(r_psi) > 2 \
          else type("T",(),{"correlation":float("nan"),"pvalue":float("nan")})()
    log(f"Kendall tau (weighted psi vs deployed lexicographic ranking) = {tau.correlation:.4f} (p={tau.pvalue:.3g})")
    save_json({"kendall_tau": float(tau.correlation), "p_value": float(tau.pvalue),
               "alpha":a, "beta":b, "gamma":g_}, "campaign_ranking_agreement.json")

    save_csv(camp_psi, "campaign_candidates.csv")
    top = camp_psi.head(5)[["campaign_id","evidence_nodes","cyber_nodes","video_nodes",
                            "hub_nodes","mean_fusion","psi"]]
    to_latex(top, "tab_campaign_candidates.tex",
        "Top campaign candidates under the weighted score $\\psi$. The deployed implementation "
        "ranked lexicographically by size; the Kendall $\\tau$ between the two orderings is reported "
        "in the run log.", "tab:campaign_candidates")
    display(top)

    plot_df = camp[(camp.nodes > 1) & (camp.nodes < 200)]
    plt.figure(figsize=(8,5)); plt.hist(plot_df.nodes, bins=30)
    plt.title("Campaign sizes, non-singleton components (2-199 nodes)")
    plt.xlabel("Campaign size (nodes)"); plt.ylabel("Count")
    save_fig("fig_campaign_size_distribution.png")
```

## 17. What this pipeline found

Produced by running this notebook against ToN-IoT (`ton-iot-train_test_network.csv`, 211,043 rows)
and the regenerated surveillance stream. These are the numbers reported in the manuscript.

**Reproduction guards.** The notebook asserts that the regenerated stream matches the submitted one
(43,200 events; hidden-incident 34,819/8,381; label 20,577/22,623), and reports whether the fusion
stage reproduces 12,961 windows / 9,524 dual / 3,437 single. Divergences are printed with a
`<-- DIFFERS` marker. A one-window difference at the boundary is benign; anything larger should be
reconciled before the output is used.

**Branches**
- Cyber, random forest: balanced accuracy **0.993**, ROC-AUC 0.998, FPR 0.008. Every paired
  contrast against the alternatives excludes zero (Holm-corrected).
- Surveillance: **logistic regression is the best model, not random forest** — the original
  selection was a single-split artefact. Balanced accuracy **0.714 [0.713, 0.715]**.
- Ablation: the three quality descriptors are worth **+0.017 to +0.020** balanced accuracy, every
  interval excluding zero, Holm p < 2e-5.
- Removing `hidden_incident` costs at most **0.003** balanced accuracy on the full feature set, and
  for logistic regression the interval includes zero: the leak was real but **not load-bearing**.

**Fusion**
- UVIF beats the cyber branch alone and mean fusion at the operating point.
- UVIF is **worse in threshold-free ranking** than mean, Dempster–Shafer and evidential fusion.
- Against confidence weighting without the variational form the margin is small and attributable to
  the confidence cap, not to the derivation.
- **Every window on which the two branches disagree is cyber-positive and surveillance-negative.**
  Both the OR target and the surveillance label are constant there, so no subset of this experiment
  can reward a fusion rule independently. The notebook detects this and reports it rather than
  failing.

**Sensitivity**
- Cyber ordering: window and dual-modality counts are **identical** under native, random and
  rate-stratified orderings. The cross-modal temporal structure carries no information.
- Cosine similarity threshold: **inert** from 0.92 to 0.999. The 5,000-alert sampling cap controls
  the outcome, moving the giant component from about 21% to 94% of nodes.
- Window size: the dual-modality count is constant at or below 120 s, because the surveillance
  stream emits one event per 2 min. The deployed 120 s is the largest non-aliasing window.
